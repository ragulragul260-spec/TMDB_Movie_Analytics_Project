```python
print(df.isnull().sum())
```

```python
print("Duplicate rows:")
print(df.duplicated().sum())
```

```python
print(df[df.duplicated(keep=False)])
```

```python
df = df.drop_duplicates()

print(df.shape)
```

```python
print("Movie ID data type:")
print(df["movie_id"].dtype)

print("\nGenre ID data type:")
print(df["genre_id"].dtype)

print("\nNegative movie IDs:")
print((df["movie_id"] < 0).sum())

print("\nNegative genre IDs:")
print((df["genre_id"] < 0).sum())

print("\nZero movie IDs:")
print((df["movie_id"] == 0).sum())

print("\nZero genre IDs:")
print((df["genre_id"] == 0).sum())
```

```python
# Step 1: Create a copy
df_cleaned = df.copy()

# Step 2: Convert ID columns to numeric
df_cleaned["movie_id"] = pd.to_numeric(
    df_cleaned["movie_id"], errors="coerce"
)

df_cleaned["genre_id"] = pd.to_numeric(
    df_cleaned["genre_id"], errors="coerce"
)

# Step 3: Remove rows with missing IDs
df_cleaned = df_cleaned.dropna(
    subset=["movie_id", "genre_id"]
)

# Step 4: Convert IDs to integers
df_cleaned["movie_id"] = df_cleaned["movie_id"].astype("int64")
df_cleaned["genre_id"] = df_cleaned["genre_id"].astype("int64")


# Step 5: Reset row numbering
df_cleaned = df_cleaned.reset_index(drop=True)

print(df_cleaned.head())
print("\nCleaned shape:", df_cleaned.shape)
print("Missing values:")
print(df_cleaned.isnull().sum())
print("Duplicates remaining:", df_cleaned.duplicated().sum())
```

```python
print("Unique movies:")
print(df_cleaned["movie_id"].nunique())
```

```python
print("Unique genres:")
print(df_cleaned["genre_id"].nunique())
```

```python
genre_counts = df_cleaned["genre_id"].value_counts().sort_index()

print(genre_counts)
```

```python
genres_per_movie = df_cleaned.groupby("movie_id")["genre_id"].count()

print(genres_per_movie.head(10))

print("\nMinimum genres:", genres_per_movie.min())
print("Maximum genres:", genres_per_movie.max())
print("Average genres:", round(genres_per_movie.mean(), 2))
```

```python
# Save the cleaned dataset
df_cleaned.to_csv(
    "movie_genres_cleaned.csv",
    index=False
)

print("Cleaning completed successfully!")
print("Final rows:", df_cleaned.shape[0])
print("Final columns:", df_cleaned.shape[1])
```

```python
print("First 5 rows:")
print(df.head())

print("\nLast 5 rows:")
print(df.tail())

print("\nRandom 5 rows:")
print(df.sample(5))

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
print("Total duplicate rows:")
print(df.duplicated().sum())
```

```python
print(df[df.duplicated(keep=False)])
```

```python
print("Actor name spaces:")
print((df["actor_name"] != df["actor_name"].str.strip()).sum())

print("Character name spaces:")
print(
    (
        df["character_name"].notna()
        & (df["character_name"] != df["character_name"].str.strip())
    ).sum()
)
```

```python
df["actor_name"] = df["actor_name"].str.strip()
df["character_name"] = df["character_name"].str.strip()
```

```python
print(df.dtypes)

print("\nNegative IDs:")
print((df[["cast_row_id", "movie_id", "person_id"]] < 0).sum())

print("\nZero IDs:")
print((df[["cast_row_id", "movie_id", "person_id"]] == 0).sum())
```

```python
for col in ["cast_row_id", "movie_id", "person_id", "cast_order"]:
    print(col, pd.to_numeric(df[col], errors="coerce").isna().sum())
```

```python
# Step 1: Create a separate copy
df_cleaned = df.copy()

# Step 2: Remove unwanted spaces from text
df_cleaned["actor_name"] = df_cleaned["actor_name"].str.strip()
df_cleaned["character_name"] = df_cleaned["character_name"].str.strip()

# Step 3: Convert ID and order columns to numeric
numeric_cols = [
    "cast_row_id",
    "movie_id",
    "person_id",
    "cast_order"
]

for col in numeric_cols:
    df_cleaned[col] = pd.to_numeric(
        df_cleaned[col], errors="coerce"
    )

# Step 4: Remove rows with missing essential numeric values
df_cleaned = df_cleaned.dropna(subset=numeric_cols)

# Step 5: Convert numeric columns to integer
for col in numeric_cols:
    df_cleaned[col] = df_cleaned[col].astype("int64")

# Step 6: Remove exact duplicate rows
df_cleaned = df_cleaned.drop_duplicates()

# Step 7: Reset row index
df_cleaned = df_cleaned.reset_index(drop=True)

# Step 8: Check the cleaned data
print("Cleaned preview:")
print(df_cleaned.head())

print("\nFinal shape:")
print(df_cleaned.shape)

print("\nMissing values:")
print(df_cleaned.isnull().sum())

print("\nDuplicates remaining:")
print(df_cleaned.duplicated().sum())
```

```python
print("Unique movies:", df_cleaned["movie_id"].nunique())
```

```python
cast_per_movie = df_cleaned.groupby("movie_id").size()

print(cast_per_movie.head(10))

print("\nMinimum cast rows:", cast_per_movie.min())
print("Maximum cast rows:", cast_per_movie.max())
print("Average cast rows:", round(cast_per_movie.mean(), 2))
```

```python
actor_counts = df_cleaned["actor_name"].value_counts()

print(actor_counts.head(10))
```

```python
# Save cleaned dataset
df_cleaned.to_csv(
    "cast_cleaned.csv",
    index=False
)

print("Cleaning completed!")
print("Final rows:", df_cleaned.shape[0])
print("Final columns:", df_cleaned.shape[1])
```

```python
text_cols = ["person_name", "job", "department"]

for col in text_cols:
    print(
        col,
        (df[col] != df[col].str.strip()).sum()
    )
```

```python
for col in text_cols:
    df[col] = df[col].str.strip()
```

```python
print(df.dtypes)

print("\nNegative IDs:")
print((df[["crew_row_id", "movie_id", "person_id"]] < 0).sum())

print("\nZero IDs:")
print((df[["crew_row_id", "movie_id", "person_id"]] == 0).sum())
```

```python
# Step 1: Create a copy
df_cleaned = df.copy()

# Step 2: Clean text columns
text_cols = ["person_name", "job", "department"]

for col in text_cols:
    df_cleaned[col] = df_cleaned[col].str.strip()

# Step 3: Validate numeric columns
numeric_cols = ["crew_row_id", "movie_id", "person_id"]

for col in numeric_cols:
    df_cleaned[col] = pd.to_numeric(
        df_cleaned[col], errors="coerce"
    )

# Step 4: Remove rows with missing essential values
df_cleaned = df_cleaned.dropna(
    subset=numeric_cols + text_cols
)

# Step 5: Convert numeric columns to integer
for col in numeric_cols:
    df_cleaned[col] = df_cleaned[col].astype("int64")

# Step 6: Remove exact duplicate rows
df_cleaned = df_cleaned.drop_duplicates()

# Step 7: Reset index
df_cleaned = df_cleaned.reset_index(drop=True)

# Step 8: Check the final result
print("First 5 cleaned rows:")
print(df_cleaned.head())

print("\nFinal shape:")
print(df_cleaned.shape)

print("\nMissing values:")
print(df_cleaned.isnull().sum())

print("\nDuplicates remaining:")
print(df_cleaned.duplicated().sum())
```

```python
print("Unique movies:", df_cleaned["movie_id"].nunique())
```

```python
print("Unique people:", df_cleaned["person_id"].nunique())
```

```python
print(df_cleaned["job"].value_counts())
```

```python
print(df_cleaned["department"].value_counts())
```

```python
# Save cleaned dataset
df_cleaned.to_csv(
    "crew_cleaned.csv",
    index=False
)

print("Cleaning completed successfully!")
print("Final rows:", df_cleaned.shape[0])
print("Final columns:", df_cleaned.shape[1])
```

```python
df_cleaned = df.copy()

df_cleaned['genre_name'] = (
    df_cleaned['genre_name'].str.strip()
)

df_cleaned['genre_id'] = pd.to_numeric(
    df_cleaned['genre_id'],
    errors='coerce'
)

df_cleaned = df_cleaned.dropna(
    subset=['genre_id', 'genre_name']
)

df_cleaned = df_cleaned.drop_duplicates()

df_cleaned = df_cleaned.drop_duplicates(
    subset=['genre_id']
)

df_cleaned = df_cleaned.drop_duplicates(
    subset=['genre_name']
)

df_cleaned = df_cleaned.reset_index(drop=True)

print(df_cleaned.head())
```

```python
print("Shape:", df_cleaned.shape)

print("\nMissing values:")
print(df_cleaned.isnull().sum())

print("\nDuplicate rows:")
print(df_cleaned.duplicated().sum())

print("\nDuplicate genre IDs:")
print(df_cleaned['genre_id'].duplicated().sum())

print("\nDuplicate genre names:")
print(df_cleaned['genre_name'].duplicated().sum())
```

```python
df_cleaned.to_csv('genres_cleaned.csv',index=False)
```
