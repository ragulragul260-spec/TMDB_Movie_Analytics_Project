```python
import pandas as pd
import mysql.connector
import os

folder = r"C:\HCL Guvi notes\TMDB Project - 1\data"

conn = mysql.connector.connect(
    host="your_host",
    user="your_username",
    password="your_new_password",
    database="movies_db",
    port=4000
)

cursor = conn.cursor()

files = {
    "cleaned_data_movies.csv": "movies",
    "cast_cleaned.csv": "cast",
    "crew_cleaned.csv": "crew",
    "genres_cleaned.csv": "genres",
    "movie_genres_cleaned.csv": "movie_genres",
    "movie_keywords.csv": "movie_keywords"
}

for file, table in files.items():

    path = os.path.join(folder, file)

    if not os.path.exists(path):
        print(f"File not found: {path}")
        continue

    print(f"\nLoading {file}")

    df = pd.read_csv(path)

    df = df.astype(object).where(pd.notna(df), None)

    cursor.execute(f"SHOW COLUMNS FROM `{table}`")
    table_columns = [row[0] for row in cursor.fetchall()]

    columns = [col for col in table_columns if col in df.columns]

    if not columns:
        print(f"No matching columns for {table}")
        continue

    col_names = ", ".join(f"`{col}`" for col in columns)
    placeholders = ", ".join(["%s"] * len(columns))

    query = f"""
        INSERT IGNORE INTO `{table}`
        ({col_names})
        VALUES ({placeholders})
    """

    data = [
        tuple(row)
        for row in df[columns].itertuples(index=False, name=None)
    ]

    if data:
        cursor.executemany(query, data)
        conn.commit()

    print(f"{table}: {cursor.rowcount} rows inserted")

cursor.close()
conn.close()

print("\nImport completed successfully!")
```

Question 1
List all movies in the Action genre released after 2015, showing title and release date.

Concept: JOIN (movies ⋈ movie_genres ⋈ genres)

```python
import mysql.connector
import pandas as pd

conn = mysql.connector.connect(
    host="your_host",
    user="your_username",
    password="your_new_password",
    database="movies_db",
    port=4000
)

query1 = """
SELECT
    m.title,
    m.release_date
FROM movies m
JOIN movie_genres mg
    ON m.movie_id = mg.movie_id
JOIN genres g
    ON mg.genre_id = g.genre_id
WHERE g.genre_name = 'Action'
AND YEAR(m.release_date) > 2015
ORDER BY m.release_date;
"""

df = pd.read_sql(query1, conn)

print(df.head(20))

conn.close()
```

Question 2

Find the top 10 highest-grossing movies along with the genre(s) each belongs to.

Concept Focus: JOIN + ORDER BY + LIMIT

```python
import mysql.connector
import pandas as pd

conn = mysql.connector.connect(
    host="your_host",
    user="your_username",
    password="your_new_password",
    database="movies_db",
    port=4000
)

query2 = """
SELECT
    m.title,
    m.revenue,
    GROUP_CONCAT(g.genre_name SEPARATOR ', ') AS genres
FROM movies m
JOIN movie_genres mg
    ON m.movie_id = mg.movie_id
JOIN genres g
    ON mg.genre_id = g.genre_id
GROUP BY m.movie_id, m.title, m.revenue
ORDER BY m.revenue DESC
LIMIT 10;
"""

df = pd.read_sql(query2, conn)

print(df.head(20))

conn.close()
```

Question 3
Calculate the average budget and average revenue for each genre, sorted by average revenue descending.

Concept: JOIN + GROUP BY + AVG()

```python
import mysql.connector
import pandas as pd

conn = mysql.connector.connect(
    host="your_host",
    user="your_username",
    password="your_new_password",
    database="movies_db",
    port=4000
)

query3 = """
SELECT
    g.genre_name,
    ROUND(AVG(m.budget), 2) AS avg_budget,
    ROUND(AVG(m.revenue), 2) AS avg_revenue
FROM movies m
JOIN movie_genres mg
    ON m.movie_id = mg.movie_id
JOIN genres g
    ON mg.genre_id = g.genre_id
GROUP BY g.genre_name
ORDER BY avg_revenue DESC;
"""

df = pd.read_sql(query3, conn)

print(df.head(20))
print(df)
```

Question 5
Find all directors who have directed more than 3 movies, along with their movie count.

Concept: GROUP BY + HAVING

```python
import mysql.connector
import pandas as pd

conn = mysql.connector.connect(
    host="your_host",
    user="your_username",
    password="your_new_password",
    database="movies_db",
    port=4000
)




query5 = """
SELECT
    person_name AS director_name,
    COUNT(DISTINCT movie_id) AS movie_count
FROM crew
WHERE job = 'Director'
GROUP BY person_name
HAVING COUNT(DISTINCT movie_id) > 3
ORDER BY movie_count DESC;
"""

df = pd.read_sql(query5, conn)

print(df)
```

```python
query7 = """
SELECT
    title,
    budget,
    revenue
FROM movies
WHERE budget >
(
    SELECT AVG(budget)
    FROM movies
)
AND revenue <
(
    SELECT AVG(revenue)
    FROM movies
)
ORDER BY budget DESC;
"""

df = pd.read_sql(query7, conn)

print(df)
```

Question 9
Find the genre with the highest average rating, considering only genres with at least 20 movies.

Concept: JOIN + GROUP BY + HAVING + AVG()

```python
query9 = """
SELECT
    g.genre_name,
    ROUND(AVG(m.vote_average), 2) AS avg_rating,
    COUNT(*) AS movie_count
FROM movies m
JOIN movie_genres mg
    ON m.movie_id = mg.movie_id
JOIN genres g
    ON mg.genre_id = g.genre_id
GROUP BY g.genre_name
HAVING COUNT(*) >= 20
ORDER BY avg_rating DESC
LIMIT 1;
"""

df = pd.read_sql(query9, conn)

print(df)
```

Question 6

```python
# Calculate Quartiles

Q1 = movies["runtime"].quantile(0.25)
Q3 = movies["runtime"].quantile(0.75)

# Calculate IQR

IQR = Q3 - Q1

# Define boundaries

lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

# Find runtime outliers

runtime_outliers = movies[
    (movies["runtime"] < lower_bound) |
    (movies["runtime"] > upper_bound)
]

runtime_outliers[
    [
        "movie_id",
        "title",
        "runtime"
    ]
].sort_values(
    by="runtime",
    ascending=False
)
```

Question 10
List movies whose popularity score and vote count do not follow the general pattern between the two measures.

```python
# Calculate popularity and vote count thresholds

popularity_q95 = movies["popularity"].quantile(0.95)
vote_count_q95 = movies["vote_count"].quantile(0.95)

# High popularity but low vote count

high_pop_low_votes = movies[
    (movies["popularity"] >= popularity_q95) &
    (movies["vote_count"] < movies["vote_count"].median())
]

# Low popularity but high vote count

low_pop_high_votes = movies[
    (movies["vote_count"] >= vote_count_q95) &
    (movies["popularity"] < movies["popularity"].median())
]

print("High Popularity, Low Vote Count")
display(
    high_pop_low_votes[
        [
            "title",
            "popularity",
            "vote_count"
        ]
    ]
)

print("\nLow Popularity, High Vote Count")
display(
    low_pop_high_votes[
        [
            "title",
            "popularity",
            "vote_count"
        ]
    ]
)
```

```python
unusual_movies = pd.concat([
    high_pop_low_votes,
    low_pop_high_votes
])

unusual_movies[
    [
        "title",
        "popularity",
        "vote_count"
    ]
].sort_values(
    by="popularity",
    ascending=False
)
```
